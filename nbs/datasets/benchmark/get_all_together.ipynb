```python
import pandas as pd

from nlp4bia.datasets.benchmark.distemist import DistemistLoader, DistemistGazetteer
from nlp4bia.datasets.benchmark.symptemist import SymptemistLoader, SymptemistGazetteer
from nlp4bia.datasets.benchmark.medprocner import MedprocnerLoader, MedprocnerGazetteer
from nlp4bia.datasets.benchmark.meddoplace import MeddoplaceLoader, MeddoplaceGazetteer
```

```python
distemist = DistemistLoader(encoding='utf-8')
symptemist = SymptemistLoader(encoding='latin1')
medprocner = MedprocnerLoader(encoding='utf-8')
meddoplace = MeddoplaceLoader(encoding='utf-8')

distemist_gaz = DistemistGazetteer(encoding='utf-8')
symptemist_gaz = SymptemistGazetteer(encoding='utf-8')
medprocner_gaz = MedprocnerGazetteer(encoding='latin-1')
meddoplace_gaz = MeddoplaceGazetteer(encoding='utf-8')
```

```python
distemist.df
```

```python
total_data = pd.concat([distemist.df, symptemist.df, medprocner.df, meddoplace.df], ignore_index=True)
```

```python
total_gaz = pd.concat([distemist_gaz.df, symptemist_gaz.df, medprocner_gaz.df], ignore_index=True)
```
