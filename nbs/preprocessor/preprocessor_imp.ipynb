```python
import hashlib
import os
import csv
import glob
import multiprocessing
from tqdm import tqdm
import pandas as pd
import sys
csv.field_size_limit(sys.maxsize)

from nlp4bia.preprocessor.HashExtractor import HashExtractor

TEST_PATH = "nbs/preprocessor/test/test_data"
```

```python

ls_files = glob.glob("nbs/preprocessor/test/test_data/test*/**/*.txt", recursive=True)
print(ls_files)
he = HashExtractor()
print(he.get_file_hash(os.path.join(TEST_PATH, "test1/test11/example1.txt")))
print(he.get_batch_hash(ls_files, os.path.join(TEST_PATH, "hashes.csv")))
```

```python
from nlp4bia.preprocessor.HashExtractor import ParallelHashExtractor
phe = ParallelHashExtractor("nbs/preprocessor/", num_processes=4)
phe.get_batch_hash(os.path.join(TEST_PATH, "test*/**/*.txt"));
```

```python
from nlp4bia.preprocessor.CSVConcatenator import CSVConcatenator
        
csvc = CSVConcatenator("nbs/preprocessor/", num_processes=2)
print(csvc.get_file_content(os.path.join(TEST_PATH, "test1/test11/example1.txt")))    
print(csvc.get_batch_content(ls_files, os.path.join(TEST_PATH, "contents.csv")))
```

```python
from nlp4bia.preprocessor.CSVConcatenator import ParallelCSVConcatenator
        
pcsv = ParallelCSVConcatenator("nbs/preprocessor/", num_processes=2)
pcsv.get_batch_content(os.path.join(TEST_PATH, "test*/**/*.txt"))
```
